# Taller 1 - API-Football - Análisis
Preguntas de análisis a partir de `equipos.parquet`, `partidos.parquet` y `clasificacion.parquet`, generados por `extractor_api.py`.

```python
import pandas as pd
pd.set_option('display.max_columns', None)

df_equipos = pd.read_parquet('equipos.parquet')
df_partidos = pd.read_parquet('partidos.parquet')
df_clasificacion = pd.read_parquet('clasificacion.parquet')

print(f"equipos.parquet: {df_equipos.shape[0]} filas, {df_equipos.shape[1]} columnas")
print(f"partidos.parquet: {df_partidos.shape[0]} filas, {df_partidos.shape[1]} columnas")
print(f"clasificacion.parquet: {df_clasificacion.shape[0]} filas, {df_clasificacion.shape[1]} columnas")
```

## 1. ¿Cuántos equipos participaron en la competición?

```python
cantidad_equipos = df_equipos['equipo_id'].nunique()
print(f"Participaron {cantidad_equipos} equipos.")
```

## 2. ¿Cuántos partidos se jugaron en cada ronda?

```python
partidos_por_ronda = (
    df_partidos.groupby('ronda')
    .size()
    .reset_index(name='cantidad_partidos')
    .sort_values('cantidad_partidos', ascending=False)
)
partidos_por_ronda
```

## 3. ¿Cuál fue el partido con mayor cantidad total de goles?

`total_goles = goles_local + goles_visitante`. Si varios partidos tienen la misma cantidad máxima, se muestran todos.

```python
df_partidos['total_goles'] = df_partidos['goles_local'] + df_partidos['goles_visitante']

max_goles = df_partidos['total_goles'].max()
partidos_mas_goleados = df_partidos[df_partidos['total_goles'] == max_goles]

partidos_mas_goleados[[
    'equipo_local_nombre', 'goles_local', 'goles_visitante', 'equipo_visitante_nombre',
    'total_goles', 'ronda', 'estadio_nombre'
]]
```

## 4. ¿Cuál fue el equipo que anotó más goles durante toda la competición?

Se suman los goles anotados como local y como visitante. Si hay empate, se muestran todos los equipos.

```python
goles_como_local = df_partidos[['equipo_local_id', 'equipo_local_nombre', 'goles_local']].rename(
    columns={'equipo_local_id': 'equipo_id', 'equipo_local_nombre': 'nombre_equipo', 'goles_local': 'goles'}
)
goles_como_visitante = df_partidos[['equipo_visitante_id', 'equipo_visitante_nombre', 'goles_visitante']].rename(
    columns={'equipo_visitante_id': 'equipo_id', 'equipo_visitante_nombre': 'nombre_equipo', 'goles_visitante': 'goles'}
)

goles_totales = pd.concat([goles_como_local, goles_como_visitante])
goles_por_equipo = goles_totales.groupby(['equipo_id', 'nombre_equipo'])['goles'].sum().reset_index()

max_goles_equipo = goles_por_equipo['goles'].max()
equipos_mas_goleadores = goles_por_equipo[goles_por_equipo['goles'] == max_goles_equipo]
equipos_mas_goleadores.sort_values('goles', ascending=False)
```

## 5. ¿Cuál fue el equipo que ganó más partidos?

Se usan las columnas `gano_local` y `gano_visitante`. Las victorias se cuentan tanto cuando el equipo jugó de local como de visitante.

```python
victorias_local = df_partidos[df_partidos['gano_local'] == True][
    ['equipo_local_id', 'equipo_local_nombre']
].rename(columns={'equipo_local_id': 'equipo_id', 'equipo_local_nombre': 'nombre_equipo'})

victorias_visitante = df_partidos[df_partidos['gano_visitante'] == True][
    ['equipo_visitante_id', 'equipo_visitante_nombre']
].rename(columns={'equipo_visitante_id': 'equipo_id', 'equipo_visitante_nombre': 'nombre_equipo'})

victorias_totales = pd.concat([victorias_local, victorias_visitante])
victorias_por_equipo = (
    victorias_totales.groupby(['equipo_id', 'nombre_equipo'])
    .size()
    .reset_index(name='partidos_ganados')
)

max_victorias = victorias_por_equipo['partidos_ganados'].max()
equipos_mas_ganadores = victorias_por_equipo[victorias_por_equipo['partidos_ganados'] == max_victorias]
equipos_mas_ganadores.sort_values('partidos_ganados', ascending=False)
```

## 6. ¿Cuál fue el equipo con la mejor diferencia de gol dentro de cada grupo?

Si varios equipos tienen la misma diferencia máxima dentro del grupo, se muestran todos.

```python
diferencia_max_grupo = df_clasificacion.groupby('grupo')['diferencia_gol'].transform('max')
mejores_diferencia = df_clasificacion[df_clasificacion['diferencia_gol'] == diferencia_max_grupo]

mejores_diferencia[['grupo', 'nombre_equipo', 'diferencia_gol']].sort_values('grupo')
```

## 7. ¿Qué equipos terminaron en la primera posición de cada grupo?

```python
primeros_de_grupo = df_clasificacion[df_clasificacion['posicion'] == 1]
primeros_de_grupo[['grupo', 'nombre_equipo', 'puntos', 'diferencia_gol']].sort_values('grupo')
```

## 8. ¿Cuál fue el estadio en el que se disputaron más partidos?

Si varios estadios tienen la misma cantidad máxima de partidos, se muestran todos.

```python
partidos_por_estadio = (
    df_partidos.groupby(['estadio_id', 'estadio_nombre', 'estadio_ciudad'])
    .size()
    .reset_index(name='cantidad_partidos')
)

max_partidos_estadio = partidos_por_estadio['cantidad_partidos'].max()
estadios_con_mas_partidos = partidos_por_estadio[
    partidos_por_estadio['cantidad_partidos'] == max_partidos_estadio
]
estadios_con_mas_partidos
```

## 9. ¿Existen partidos duplicados según el campo `partido_id`?

```python
cantidad_duplicados = df_partidos['partido_id'].duplicated().sum()

if cantidad_duplicados > 0:
    print(f"Sí, hay {cantidad_duplicados} registros duplicados según partido_id.")
    display(df_partidos[df_partidos['partido_id'].duplicated(keep=False)].sort_values('partido_id'))
else:
    print("No hay partidos duplicados según partido_id.")
```

## 10. ¿Cuántos valores nulos tiene cada columna de cada archivo?

```python
archivos = {
    'equipos.parquet': df_equipos,
    'partidos.parquet': df_partidos,
    'clasificacion.parquet': df_clasificacion,
}

for nombre_archivo, df in archivos.items():
    print(f"\n--- {nombre_archivo} ---")
    nulos = df.isnull().sum()
    nulos = nulos[nulos > 0] if nulos.sum() > 0 else nulos
    print(nulos.to_string() if len(nulos) > 0 else "Sin valores nulos en ninguna columna.")
```
